```python
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
stats = []
error = []
std = []
```

Mean signed error (bias): mean(pred - truth). Positive = overprediction on average, negative = underprediction. Your primary metric for the "does the model swing over/under across seasons" question.

Standard deviation of signed errors: std(pred - truth). Tells you the spread of errors around the bias. If bias is zero but SD is large, the model is unbiased on average but wildly wrong per tract. If bias is nonzero and SD is small, the model is systematically off in one direction. These are different diagnostic pictures.

RMSE: sqrt(mean((pred - truth)^2)). Combines bias and spread into one number. Useful for reporting alongside R² but doesn't distinguish over/underprediction.

```python
df = pd.read_csv("/data/hbaier/new_data/tlag/az_imagery/q1_2016_s2_allbands/artifacts/az_q1_2016_allbands_v1/epoch174_valset_preds.csv")
print(r2_score(df["label"], df["pred"]))
stats.append(r2_score(df["label"], df["pred"]))
error.append(np.mean(df["pred"] - df["label"]))
std.append(np.std(df["pred"] - df["label"]))
df.head()
```
